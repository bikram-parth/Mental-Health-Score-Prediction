```python
Countries=['India','USA','Canada','Australia','UK','Germany','Mexico','Turkey','France']
def group_countries(country):
    if country not in Countries:
        return 'other'
    else:
        return country
```

```python
from sklearn.preprocessing import FunctionTransformer,StandardScaler,OrdinalEncoder,OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
df_cleaned.head()
```

```python
df_cleaned.columns
```

```python
skewed_col=['Study_Hours']
ordinal_col=['Stress_Level']
one_hot_col=['Gender', 'Country', 'Academic_Level', 'Most_Used_Platform',
       'Purpose_Of_Use',
       'Stress_Level']
normal_col=['Age','Avg_Daily_Usage_Hours', 'Daily_Unlocks',
       'Physical_Activity_Hours', 'Sleep_Hours_Per_Night',
       ]
feature_col=skewed_col+ordinal_col+one_hot_col+normal_col
X=df[feature_col]
y=df['Mental_Health_Score']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.30,random_state=42)
```

```python
from sklearn.pipeline import Pipeline
```

```python
skew_pipeline=Pipeline([
    ('log_transform',FunctionTransformer(np.log1p)),
    ('standard_scalar',StandardScaler())
])

ordinal_pipeline=Pipeline([
    ('ordinal',OrdinalEncoder(categories=[['Low', 'Medium', 'High', 'Very High']])),
])

one_hot_pipeline=Pipeline([
    ('one_hot',OneHotEncoder(handle_unknown='ignore')),
])

normal_pipeline=Pipeline([
    ('scalar',StandardScaler())
])
```

```python
prepocessor=ColumnTransformer(transformers=[
    ('skew_process',skew_pipeline,skewed_col),
    ('ordinal_process',ordinal_pipeline,ordinal_col),
    ('one_hot_encode',one_hot_pipeline,one_hot_col),
    ('normal_pipeline',normal_pipeline,normal_col)
])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import AdaBoostRegressor
from sklearn.ensemble import GradientBoostingRegressor
from xgboost import XGBRegressor
```

```python
linear_model=Pipeline(steps=[
    ('preprocessor',prepocessor),
    ('Linear_model',LinearRegression())
])
```

```python
X_train = X_train.loc[:, ~X_train.columns.duplicated()]
X_test = X_test.loc[:, ~X_test.columns.duplicated()]
```

```python
X_train.columns
```

```python
linear_model.fit(X_train,y_train)
```

**testing score of linear model**

```python
y_pred_linear=linear_model.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
```

```python
r2_score(y_test,y_pred_linear)
```

**training score of linear model**

```python
y_pred_train_linear=linear_model.predict(X_train)
```

```python
r2_score(y_train,y_pred_train_linear)
```

```python
rf_pipeline=Pipeline(steps=[
    ('prepocessor',prepocessor),
    ('random_model',RandomForestRegressor(n_estimators=100))
])
```

```python
rf_pipeline.fit(X_train,y_train)
```

```python
y_pred_rf_train=rf_pipeline.predict(X_train)
y_pred_rf_test=rf_pipeline.predict(X_test)
```

```python
print("train accuracy of randomforest_model : ",r2_score(y_train,y_pred_rf_train))
print("test accuracy of randomforest_model : ",r2_score(y_test,y_pred_rf_test))
```

**performing hyperparamter tuning on RandomForest model**

```python
from sklearn.model_selection import GridSearchCV,RandomizedSearchCV
```

```python
Random_search_rf=RandomizedSearchCV(
    rf_pipeline,
    param_distributions={
        'random_model__n_estimators': [100, 200, 300],
        'random_model__max_depth': [None, 10, 20, 30],
        'random_model__min_samples_split': [2, 5, 10],
        'random_model__min_samples_leaf': [1, 2, 4],
        'random_model__max_features': ['sqrt', 'log2', None]
    },
    n_iter=5,
    cv=5
)
```

```python
Random_search_rf.fit(X_train,y_train)
```

```python
y_pred_test_rf=Random_search_rf.predict(X_test)
```

```python
y_pred_train_rf=Random_search_rf.predict(X_train)
```

```python
y_pred_train_rf
```

```python
print("train accuracy of randomforest_model : ",r2_score(y_train,y_pred_train_rf))
print("test accuracy of randomforest_model : ",r2_score(y_test,y_pred_test_rf))
```
